```python
import math, numpy as np, matplotlib.pyplot as plt
print("=== ANALIZADOR IR UNCP - INTERACTIVO ===")
wn = float(input("Ingresa numero de onda (ej 1715): "))
dx = float(input("Dipolo x (ej 0.8): "))
dy = float(input("Dipolo y (ej 0.5): "))
dz = float(input("Dipolo z (ej 0.3): "))
ex = float(input("Campo E x (ej 1): "))
ey = float(input("Campo E y (ej 0.2): "))
ez = float(input("Campo E z (ej 0.1): "))
C=299792458.0; H=6.62607015e-34; NA=6.02214076e23
dm = np.array([dx,dy,dz])*1e-30
ed = np.array([ex,ey,ez]); ed=ed/np.linalg.norm(ed)
prod = np.dot(dm,ed); mag = np.linalg.norm(dm)
ang = math.degrees(math.acos(prod/mag))
f = C*wn*100; E = (H*f*NA)/1000
print(f"\n>>> RESULTADOS: {wn} cm-1 -> {E:.3f} kJ/mol")
print(f">>> Angulo: {ang:.2f} grados")
x = np.linspace(4000,400,1000)
g = lambda x,mu,s,a: a*np.exp(-0.5*((x-mu)/s)**2)
y = g(x,wn,20,1)+g(x,2950,30,0.6)+g(x,3300,40,0.4)
plt.figure(figsize=(10,4))
plt.plot(x,y,color='darkred',lw=2)
plt.axvline(wn,color='blue',linestyle='--',label=f'{wn} cm-1')
plt.gca().invert_xaxis()
plt.title(f"Espectro IR UNCP - {wn} cm-1 - {ang:.1f}°")
plt.xlabel('cm-1'); plt.ylabel('Absorbancia')
plt.legend(); plt.grid(True,alpha=0.3); plt.show()
```
